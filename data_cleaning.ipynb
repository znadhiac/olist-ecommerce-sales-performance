# **OLIST BRAZILIAN E-COMMERCE SALES PERFORMANCE DATA CLEANING**

## **1. Load Data**

```python
import pandas as pd
import numpy as np

# Fact tables
orders_raw = pd.read_csv("olist_orders_dataset.csv")
order_items_raw = pd.read_csv("olist_order_items_dataset.csv")
order_payments_raw = pd.read_csv("olist_order_payments_dataset.csv")
order_reviews_raw = pd.read_csv("olist_order_reviews_dataset.csv")

orders = orders_raw.copy()
order_items = order_items_raw.copy()
order_payments = order_payments_raw.copy()
order_reviews = order_reviews_raw.copy()

# Dimension tables
customers_raw = pd.read_csv("olist_customers_dataset.csv")
products_raw = pd.read_csv("olist_products_dataset.csv")
sellers_raw = pd.read_csv("olist_sellers_dataset.csv")
category_translation_raw = pd.read_csv("product_category_name_translation.csv", encoding="utf-8-sig")

customers = customers_raw.copy()
products = products_raw.copy()
sellers = sellers_raw.copy()
category_translation = category_translation_raw.copy()
```

## **2. Initial Data Inspection**

```python
# INITIAL DATA INSPECTION

datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "products": products,
    "customers": customers,
    "sellers": sellers,
    "category_translation": category_translation
}

for name, df in datasets.items():
    print(f"\n{'='*40}")
    print(f"{name} Table")
    print(f"{'='*40}")
    print("Shape:", df.shape)
    print("\nInfo:")
    df.info()
    print("\nPreview:")
    display(df.head())
```

## **3. Converting Incorrect Data Type**

```python
# CONVERT INCORRECT OBJECT TYPE INTO DATE TYPE

# Convert timestamps
orders["order_purchase_timestamp"] = pd.to_datetime(
    orders["order_purchase_timestamp"], errors="coerce"
)
orders["order_approved_at"] = pd.to_datetime(
    orders["order_approved_at"], errors="coerce"
)
orders["order_delivered_carrier_date"] = pd.to_datetime(
    orders["order_delivered_carrier_date"], errors="coerce"
)
orders["order_delivered_customer_date"] = pd.to_datetime(
    orders["order_delivered_customer_date"], errors="coerce"
)
orders["order_estimated_delivery_date"] = pd.to_datetime(
    orders["order_estimated_delivery_date"], errors="coerce"
)
order_items["shipping_limit_date"] = pd.to_datetime(
    order_items["shipping_limit_date"], errors="coerce"
)
order_reviews["review_creation_date"] = pd.to_datetime(
    order_reviews["review_creation_date"], errors="coerce"
)
order_reviews["review_answer_timestamp"] = pd.to_datetime(
    order_reviews["review_answer_timestamp"], errors="coerce"
)

# Keep only 2018 orders
orders = orders[
    (orders["order_purchase_timestamp"] >= "2018-01-01") &
    (orders["order_purchase_timestamp"] < "2019-01-01")]
```

## **4. Handling Missing Values**

```python
# DOUBLE CHECK MISSING VALUES FOR ALL TABLES

datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "products": products,
    "customers": customers,
    "sellers": sellers,
    "category_translation": category_translation
}

for name, df in datasets.items():
    print(f"\n{name} - Missing Values by Column:")
    print(df.isnull().sum())
```

```python
# DROP NULL VALUES

# Drop rows with nulls in products table
products = products.dropna(subset=[
    'product_id', 
    'product_category_name', 
    'product_name_lenght', 
    'product_description_lenght',
    'product_photos_qty',
    'product_weight_g',
    'product_length_cm',
    'product_height_cm',
    'product_width_cm'
])

products = products.reset_index(drop=True)

# Drop rows with nulls in orders table
orders = orders.dropna(subset=[
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date"
])

orders = orders.reset_index(drop=True)
```

```python
# DROP COLUMNS WITH PREDOMINANTLY NULL VALUES

order_reviews = order_reviews.drop(columns=[
    "review_comment_title", 
    "review_comment_message"
])
```

```python
# CHECK IF NULL VALUE STILL EXIST
datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "products": products,
    "customers": customers,
    "sellers": sellers,
    "category_translation": category_translation
}

for name, df in datasets.items():
    print(f"Nulls in {name}:")
    print(df.isnull().sum())
    print("-"*50)
```

## **5. Handling Duplicate Data**

```python
# CHECK DUPLICATE DATA FOR ALL TABLES

datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "products": products,
    "customers": customers,
    "sellers": sellers,
    "category_translation": category_translation
}

for name, df in datasets.items():
    duplicates = df.duplicated().sum()
    print(f"{name} - Number of duplicate rows: {duplicates}")
```

No duplicate entries were detected in the dataset, so no cleaning was necessary.

## **6. Handling Inconsistent Data**

```python
# CHECK CATEGORICAL VALUE CONSISTENCY FOR ALL TABLES

datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "products": products,
    "customers": customers,
    "sellers": sellers,
    "category_translation": category_translation
}

print("=== CATEGORICAL VALUE CONSISTENCY CHECK ===")
for name, df in datasets.items():
    cat_cols = df.select_dtypes(include="object").columns
    if len(cat_cols) == 0:
        print(f"\n{name}: No categorical columns")
        continue

    print(f"\n{name}:")
    for col in cat_cols:
        unique_values = df[col].str.strip().unique()
        print(f"  {col}: {unique_values}")
```

```python
# Fix typos
products = products.rename(columns={
    "product_name_lenght": "product_name_length",
    "product_description_lenght": "product_description_length"
})
```

```python
# Correct category translation column
category_translation = category_translation.rename(
    columns={"ï»¿product_category_name": "product_category_name"}
)
```

```python
# Keep review_score between 1 and 5
order_reviews = order_reviews[order_reviews['review_score'].between(1, 5)]
```

## **7. Join Validation**

```python
# CHECK FOREIGN KEYS – OLIST E-COMMERCE

# 1. ORDERS → CUSTOMERS
print("\n=== FOREIGN KEY CHECK: ORDERS → CUSTOMERS ===")
invalid_orders_customers = (~orders["customer_id"].isin(customers["customer_id"])).sum()
print(f"Orders with invalid customer_id references: {invalid_orders_customers}")

# 2. ORDER ITEMS → ORDERS
print("\n=== FOREIGN KEY CHECK: ORDER ITEMS → ORDERS ===")
invalid_orderitems_orders = (~order_items["order_id"].isin(orders["order_id"])).sum()
print(f"Order items with invalid order_id references: {invalid_orderitems_orders}")

# 3. ORDER ITEMS → PRODUCTS
print("\n=== FOREIGN KEY CHECK: ORDER ITEMS → PRODUCTS ===")
invalid_orderitems_products = (~order_items["product_id"].isin(products["product_id"])).sum()
print(f"Order items with invalid product_id references: {invalid_orderitems_products}")

# 4. ORDER ITEMS → SELLERS
print("\n=== FOREIGN KEY CHECK: ORDER ITEMS → SELLERS ===")
invalid_orderitems_sellers = (~order_items["seller_id"].isin(sellers["seller_id"])).sum()
print(f"Order items with invalid seller_id references: {invalid_orderitems_sellers}")

# 5. ORDER PAYMENTS → ORDERS
print("\n=== FOREIGN KEY CHECK: ORDER PAYMENTS → ORDERS ===")
invalid_payments_orders = (~order_payments["order_id"].isin(orders["order_id"])).sum()
print(f"Payments with invalid order_id references: {invalid_payments_orders}")

# 6. ORDER REVIEWS → ORDERS
print("\n=== FOREIGN KEY CHECK: ORDER REVIEWS → ORDERS ===")
invalid_reviews_orders = (~order_reviews["order_id"].isin(orders["order_id"])).sum()
print(f"Reviews with invalid order_id references: {invalid_reviews_orders}")
```

```python
# HANDLING ORPHAN VALUE

orders = orders.dropna(subset=[
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]).copy()

# Keep only rows that reference valid orders
valid_order_ids = orders["order_id"]
order_items = order_items[order_items["order_id"].isin(valid_order_ids)].copy()
order_payments = order_payments[order_payments["order_id"].isin(valid_order_ids)].copy()
order_reviews = order_reviews[order_reviews["order_id"].isin(valid_order_ids)].copy()

# Keep only order_items that reference existing products
order_items = order_items[order_items["product_id"].isin(products["product_id"])].copy()
```

```python
# RE-CHECK HANDLED ORPHAN VALUES

# 1. ORDERS → CUSTOMERS
print("\n=== FOREIGN KEY CHECK: ORDERS → CUSTOMERS ===")
invalid_orders_customers = (~orders["customer_id"].isin(customers["customer_id"])).sum()
print(f"Orders with invalid customer_id references: {invalid_orders_customers}")

# 2. ORDER ITEMS → ORDERS
print("\n=== FOREIGN KEY CHECK: ORDER ITEMS → ORDERS ===")
invalid_orderitems_orders = (~order_items["order_id"].isin(orders["order_id"])).sum()
print(f"Order items with invalid order_id references: {invalid_orderitems_orders}")

# 3. ORDER ITEMS → PRODUCTS
print("\n=== FOREIGN KEY CHECK: ORDER ITEMS → PRODUCTS ===")
invalid_orderitems_products = (~order_items["product_id"].isin(products["product_id"])).sum()
print(f"Order items with invalid product_id references: {invalid_orderitems_products}")

# 4. ORDER ITEMS → SELLERS
print("\n=== FOREIGN KEY CHECK: ORDER ITEMS → SELLERS ===")
invalid_orderitems_sellers = (~order_items["seller_id"].isin(sellers["seller_id"])).sum()
print(f"Order items with invalid seller_id references: {invalid_orderitems_sellers}")

# 5. ORDER PAYMENTS → ORDERS
print("\n=== FOREIGN KEY CHECK: ORDER PAYMENTS → ORDERS ===")
invalid_payments_orders = (~order_payments["order_id"].isin(orders["order_id"])).sum()
print(f"Payments with invalid order_id references: {invalid_payments_orders}")

# 6. ORDER REVIEWS → ORDERS
print("\n=== FOREIGN KEY CHECK: ORDER REVIEWS → ORDERS ===")
invalid_reviews_orders = (~order_reviews["order_id"].isin(orders["order_id"])).sum()
print(f"Reviews with invalid order_id references: {invalid_reviews_orders}")
```

## **8. Final Data Inspection**

```python
# FINAL DATA INSPECTION

datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_payments": order_payments,
    "order_reviews": order_reviews,
    "products": products,
    "customers": customers,
    "sellers": sellers,
    "category_translation": category_translation
}

for name, df in datasets.items():
    print(f"\n{'='*40}")
    print(f"{name} Table")
    print(f"{'='*40}")
    print("Shape:", df.shape)
    print("\nInfo:")
    df.info()
    print("\nPreview:")
    display(df.head())
```

```python
orders.to_csv("orders.csv", index=False)
order_items.to_csv("order_items.csv", index=False)
order_payments.to_csv("order_payments.csv", index=False)
order_reviews.to_csv("order_reviews.csv", index=False)
products.to_csv("products.csv", index=False)
customers.to_csv("customers.csv", index=False)
sellers.to_csv("sellers.csv", index=False)
category_translation.to_csv("category_translation.csv", index=False)
```
